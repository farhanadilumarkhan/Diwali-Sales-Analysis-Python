```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Diwali Sales Data.csv', encoding='unicode_escape')
```

```python
df.head()
```

```python
df.info()
```

```python
# drop unrelated & blank columns
df.drop(['Status' , 'unnamed1'] , axis = 1, inplace=True)
```

```python
pd.isnull(df).sum()
```

```python
df.dropna(inplace=True)
```

```python
df['Amount'] = df['Amount'].astype('int')
```

```python
df['Amount'].dtypes
```

```python
df.columns
```

```python
# Rename column
df.rename(columns={'Cust_name' : 'Customer_Name'})
```

```python
# Use describe method for specific columns
df[['Age' , 'Orders' , 'Amount']].describe()
```

# Exploratory Data Analysis

```python
df.columns
```

```python
ax = sns.countplot(x = 'Gender' , data = df)
for bar in ax.containers:
    ax.bar_label(bar)
```

```python
sales_dist = df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x='Gender' , y='Amount' , data=sales_dist )
```

From above graph we can see that the sales amount is higher for female customers as compared to male customers.    

### Age

```python
ax = sns.countplot(x='Age Group' , data=df , hue='Gender')
for bar in ax.containers:
    ax.bar_label(bar)
```

```python
# Amount vs Age Group
sales_vs_age = df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x='Age Group', y='Amount', data=sales_vs_age)
```

From above graph we can see that most of the buyers are age group of (26 - 35) female

## States

```python
state_orders = df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)
sns.set(rc={'figure.figsize' : (15, 8)})
sns.barplot(x='State' , y='Orders' , data=state_orders)
```

```python
# Total Sales from top 10 states
sales_cities = df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)
sns.set(rc={'figure.figsize' : (15, 8)})
sns.barplot(x='State' , y='Amount' , data=sales_cities)
```

From above graph we can see that most of the orders are from Uttar Pradesh, Maharashtra, Karnataka respectively, but total sales is from UP, Karnataka & Maharashtra

## Marital Status

```python
ax = sns.countplot(data = df , x='Marital_Status')
sns.set(rc={'figure.figsize' : (5, 2)})
for bar in ax.containers:
    ax.bar_label(bar)
```

```python
sales = df.groupby(['Marital_Status' , 'Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.set(rc={'figure.figsize' : (6, 5)})
sns.barplot(x='Marital_Status', y='Amount', data=sales, hue='Gender')
```

From above graph we can see that most of the buyers are married women and they have high purchasing power

## Occupation

```python
sns.set(rc={'figure.figsize' : (20, 10)})
oc =sns.countplot(data = df, x='Occupation')
for bar in oc.containers:
    oc.bar_label(bar)
```

```python
occ_sales = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x='Occupation', y='Amount', data=occ_sales)
```

Most of the buyers are working in It Scetor, Healthcare & Aviation.

## Product Category

```python
sns.set(rc={'figure.figsize' : (25, 10)})
prod = sns.countplot(data=df, x='Product_Category')
for bar in prod.containers:
    prod.bar_label(bar)
```

```python
sns.set(rc={'figure.figsize' : (20, 5)})
prod_sales = df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)
sns.barplot(x='Product_Category', y='Amount', data=prod_sales)
```

Most of the sold product are from Food, Clothing and Electronics categories.

```python
# Top 10 Product with highest sales
pro_sale = df .groupby(['Product_ID'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)
sns.barplot(data=pro_sale, x='Product_ID', y='Amount')
```

```python
# Top 10 most sold products
fig1, ax1 = plt.subplots(figsize=(15, 5))
df.groupby('Product_ID')['Orders'].sum().nlargest(10).sort_values(ascending=False).plot(kind='bar')
```

## Conclusion:

Women aged 26-35 from Uttar Pradesh, Maharashtra and Karnataka, working in IT, Healthcare and Aviation, are the most likely to buy during Diwali, mainly from the Food, Clothing and Electronics categories.
